## Latihan 1: Operasi Dasar NumPy Array

```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

## Tugas Analisis 1: Perbandingan NumPy Array dan List Python

Untuk membandingkan hasil `harga * 0.9` dengan list Python biasa, mari kita coba operasi serupa pada list Python dan lihat apa yang terjadi. Setelah itu, akan dijelaskan mengapa perilaku keduanya berbeda.

```python
list_harga = [5000, 7000, 3000]

# Mencoba operasi perkalian pada list Python
try:
    print('Mencoba perkalian pada list:', list_harga * 0.9)
except TypeError as e:
    print(f'Error saat mencoba perkalian pada list: {e}')

print('''\nPenjelasan:\n1.  **NumPy Array**: Ketika Anda mengalikan array NumPy dengan skalar (seperti 0.9), operasi ini dilakukan secara element-wise, yaitu setiap elemen dalam array dikalikan dengan skalar tersebut. Ini adalah perilaku yang diinginkan untuk operasi matematis pada seluruh dataset.\n2.  **List Python Biasa**: List Python tidak mendukung operasi perkalian element-wise dengan skalar. Ketika Anda mencoba mengalikan list dengan sebuah integer `n`, list tersebut akan direplikasi `n` kali (misalnya `[1,2]*3` akan menjadi `[1,2,1,2,1,2]`). Namun, ketika Anda mencoba mengalikan list dengan float (seperti 0.9), Python akan menghasilkan `TypeError` karena operasi tersebut tidak didefinisikan untuk tipe data list dan float secara langsung.''')
```

## Latihan 4: Menangani Missing Value

```python
print('Jumlah data kosong tiap kolom sebelum penanganan:')
print(df.isnull().sum())

df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong

print('\nJumlah data kosong tiap kolom setelah penanganan:')
print(df.isnull().sum())
print('\nDataFrame setelah penanganan missing values:')
print(df)
```

## Tugas Analisis 4: Mengapa `fillna` dan `dropna`?

Mengapa pada contoh di atas kolom `terjual` diisi (`fillna`) sedangkan baris dengan `menu` kosong justru dihapus (`dropna`)? Diskusikan alasan logisnya.

**Alasan Logis:**
*   **Kolom `terjual` diisi (`fillna(0)`)**: Kolom `terjual` adalah data kuantitatif. Nilai `None` di sini kemungkinan besar berarti 'tidak ada penjualan' atau '0 unit terjual'. Mengisi dengan 0 adalah asumsi yang wajar dan seringkali tidak bias, memungkinkan perhitungan agregat (seperti total penjualan) tetap akurat tanpa menghilangkan data penting lainnya dari baris tersebut.
*   **Baris `menu` kosong dihapus (`dropna(subset=['menu'])`)**: Kolom `menu` adalah identitas atau kategori utama dari setiap transaksi. Jika `menu` kosong, kita kehilangan informasi fundamental tentang apa yang dijual. Mengisi dengan nilai arbitrer (misalnya 'Unknown') bisa menyesatkan analisis. Oleh karena itu, menghapus baris yang kehilangan informasi inti ini lebih aman untuk menjaga integritas data.

**Dugaan Masalah Kualitas Data:**
- Terdapat *missing value* (data kosong) pada jumlah terjual atau harga.
- Kemungkinan ada baris transaksi yang terinput ganda (duplikat).
- Tipe data harga mungkin terbaca sebagai teks (string) alih-alih angka (integer).

```python
# 1. Import library standar untuk komputasi numerik dan manipulasi data
import pandas as pd
import numpy as np

# 2. Proses Data Loading
# Ganti 'dataset_kantin.csv' dengan nama file asli kelompok Anda
df = pd.read_csv('/content/data_kantin.csv')

# Menampilkan 5 baris pertama untuk verifikasi visual
display(df.head())
```

```python
print("=== STRUKTUR DATA (SHAPE) ===")
# Melihat dimensi (jumlah baris, jumlah kolom)
print(df.shape)
print("\n")

print("=== INFORMASI DATA & TIPE DATA (INFO) ===")
# Mendeteksi keberadaan nilai Null dan memastikan tipe data tiap kolom
df.info()
print("\n")

print("=== STATISTIK DESKRIPTIF (DESCRIBE) ===")
# Menghitung metrik statistik dasar (min, max, mean) untuk kolom numerik
display(df.describe())
```

```python
# 1. Menangani Data Duplikat[cite: 9]
jumlah_duplikat = df.duplicated().sum()
print(f"Ditemukan {jumlah_duplikat} baris duplikat.")

# Alasan: Baris duplikat akan merusak kalkulasi total pendapatan, sehingga wajib dihapus.
df = df.drop_duplicates()

# 2. Menangani Missing Value (Data Kosong)[cite: 9]
print("\nJumlah Missing Value per kolom:")
print(df.isnull().sum())

# Skenario A: Jika kolom 'menu' kosong, kita hapus barisnya (Dropna)
# Alasan: Transaksi tanpa nama barang tidak valid dan tidak bisa dianalisis.
df = df.dropna(subset=['menu'])

# Skenario B: Jika kolom 'terjual' kosong, kita isi dengan angka 0 (Fillna)
# Alasan: Kolom kosong pada jumlah terjual diasumsikan hari itu tidak ada barang yang laku, bukan error sistem.
df['terjual'] = df['terjual'].fillna(0)

# 3. Standardisasi Tipe Data[cite: 9]
# Alasan: Memastikan kolom harga dan terjual bertipe integer agar bisa dikalkulasi matematis secara presisi.
df['harga'] = df['harga'].astype(int)
df['terjual'] = df['terjual'].astype(int)

# Verifikasi hasil sanitasi
print("\nStatus Data Setelah Cleaning:")
df.info()
```

```python
# 1. Membuat Kolom Turunan (Feature Engineering)[cite: 9]
# Mengkalkulasi total pendapatan per baris transaksi
df['total_pendapatan'] = df['harga'] * df['terjual']

# 2. Filtering (Penyaringan)[cite: 9]
# Hanya mengambil menu yang laku lebih dari 10 porsi agar analisis fokus pada menu potensial
df_laris = df[df['terjual'] > 10]

# 3. Groupby & Agregasi (Pengelompokan)[cite: 9]
# Menghitung total porsi terjual dan total pendapatan untuk setiap menu
ringkasan_menu = df_laris.groupby('menu').agg(
    total_porsi_terjual=('terjual', 'sum'),
    total_uang_masuk=('total_pendapatan', 'sum')
).reset_index()

# 4. Sorting (Pengurutan)[cite: 9]
# Mengurutkan data berdasarkan pendapatan tertinggi ke terendah (Descending)
hasil_akhir = ringkasan_menu.sort_values(by='total_uang_masuk', ascending=False)

display(hasil_akhir)
```

```python
# Menyimpan dataframe yang sudah bersih ke format CSV baru tanpa menyertakan indeks[cite: 9]
df.to_csv('dataset_kantin_bersih.csv', index=False)
print("Data bersih berhasil disimpan sebagai 'dataset_kantin_bersih.csv'")
```
